## Задание 1
Написать функцию для вычисления факториала числа.

```python
def factorial(n):
    """Вычисляет факториал числа n."""
    try:
        n = int(n)
        if n < 0:
            raise ValueError("Факториал определён только для неотрицательных чисел")
        result = 1
        for i in range(2, n + 1):
            result *= i
        return result
    except (TypeError, ValueError) as e:
        print(f"Ошибка: {e}")
        return None
    finally:
        print("Вычисление факториала завершено.")


print("5! =", factorial(5))
print("0! =", factorial(0))
print("(-3)! =", factorial(-3))
print("'abc'! =", factorial("abc"))
```

## Задание 2
Функция принимает один аргумент:
- **список** — сумма отрицательных чисел; удалить повторяющиеся элементы;
- **множество** — количество слов;
- **число** — простое или нет;
- **строка** — гласные и согласные; самое длинное слово.

Проверка со всеми случаями.

```python
def process_arg(value):
    """Обрабатывает аргумент в зависимости от его типа."""
    try:
        if isinstance(value, list):
            neg_sum = sum(x for x in value if isinstance(x, (int, float)) and x < 0)
            unique = list(dict.fromkeys(value))  # без повторов, порядок сохраняется
            value[:] = unique  # изменяем исходный список
            print(f"Список. Сумма отрицательных: {neg_sum}")
            print(f"Список без повторов: {value}")
            return neg_sum, value

        elif isinstance(value, set):
            words_count = sum(1 for item in value if isinstance(item, str) and item.strip())
            print(f"Множество. Количество слов: {words_count}")
            return words_count

        elif isinstance(value, bool):
            # bool — подкласс int, обрабатываем отдельно
            raise TypeError("Булево значение не поддерживается")

        elif isinstance(value, (int, float)):
            n = int(value)
            if n < 2:
                is_prime = False
            else:
                is_prime = all(n % i != 0 for i in range(2, int(n ** 0.5) + 1))
            print(f"Число {n}: {'простое' if is_prime else 'не простое'}")
            return is_prime

        elif isinstance(value, str):
            vowels_set = set("аеёиоуыэюяaeiouy")
            consonants_set = set("бвгджзйклмнпрстфхцчшщbcdfghjklmnpqrstvwxz")
            vowels = consonants = 0
            for ch in value.lower():
                if ch in vowels_set:
                    vowels += 1
                elif ch in consonants_set:
                    consonants += 1
            words = value.split()
            longest = max(words, key=len) if words else ""
            print(f"Строка. Гласных: {vowels}, согласных: {consonants}")
            print(f"Самое длинное слово: '{longest}'")
            return vowels, consonants, longest

        else:
            raise TypeError(f"Неподдерживаемый тип: {type(value).__name__}")

    except Exception as e:
        print(f"Ошибка при обработке: {e}")
        return None
    finally:
        print("Обработка аргумента завершена.\n")


# --- Проверка со всеми случаями ---
print("=== Список ===")
lst = [1, -2, 3, -4, 1, -2, 5, 3]
process_arg(lst)

print("=== Множество ===")
process_arg({"яблоко", "груша", "киви", "банан"})

print("=== Число (простое) ===")
process_arg(17)

print("=== Число (не простое) ===")
process_arg(18)

print("=== Строка ===")
process_arg("Программирование на языке Python интересно")
```

## Задание 3
Проверить, все ли строки матрицы содержат хотя бы один положительный элемент.  
Если да — изменить знаки всех элементов матрицы на обратные.

```python
def invert_matrix_if_all_rows_positive(matrix):
    """Если в каждой строке есть положительный элемент — меняет знаки всех элементов."""
    try:
        if not matrix or not all(isinstance(row, (list, tuple)) for row in matrix):
            raise ValueError("Матрица должна быть непустым списком строк")

        all_rows_ok = all(any(x > 0 for x in row) for row in matrix)

        if all_rows_ok:
            for i in range(len(matrix)):
                for j in range(len(matrix[i])):
                    matrix[i][j] = -matrix[i][j]
            print("Во всех строках есть положительный элемент. Знаки изменены.")
        else:
            print("Не во всех строках есть положительный элемент. Матрица без изменений.")

        return matrix

    except Exception as e:
        print(f"Ошибка: {e}")
        return None
    finally:
        print("Проверка матрицы завершена.")


matrix1 = [
    [1, -2, 0],
    [-5, 3, -1],
    [0, 0, 4],
]
print("Исходная матрица 1:")
for row in matrix1:
    print(row)

result1 = invert_matrix_if_all_rows_positive(matrix1)
print("Результат:")
for row in result1:
    print(row)

print()

matrix2 = [
    [1, 2],
    [-3, -4],  # нет положительных
]
print("Исходная матрица 2:")
for row in matrix2:
    print(row)

result2 = invert_matrix_if_all_rows_positive(matrix2)
print("Результат:")
for row in result2:
    print(row)
```

## Задание 4
Программа, демонстрирующая работу `try` / `except` / `finally`.

```python
def demo_try_except_finally():
    """Демонстрация try / except / finally: деление двух чисел."""
    print("=== Демонстрация try / except / finally ===\n")

    try:
        a = float(input("Введите делимое: "))
        b = float(input("Введите делитель: "))
        result = a / b
        print(f"Результат деления: {result}")
    except ZeroDivisionError:
        print("Ошибка: деление на ноль!")
    except ValueError:
        print("Ошибка: введено не число!")
    except Exception as e:
        print(f"Непредвиденная ошибка: {e}")
    finally:
        print("Блок finally: выполняется всегда (успех или ошибка).")


demo_try_except_finally()
```
