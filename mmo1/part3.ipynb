## Задание 1
Класс `Example`: 3 метода, 2 статические и 2 динамические переменные.

```python
a = 10  # глобальная переменная
b = 20  # глобальная переменная


class Example:
    # две статические (классовые) переменные
    static_x = 5
    static_y = 7

    def __init__(self, dyn1, dyn2):
        # две динамические (экземплярные) переменные
        self.dyn1 = dyn1
        self.dyn2 = dyn2

    def method1(self):
        """Создаёт переменную и выводит её."""
        local_var = "локальная переменная метода 1"
        print(local_var)

    def method2(self):
        """Возвращает сумму двух глобальных переменных."""
        return a + b

    def method3(self):
        """Возвращает dyn1 ** dyn2."""
        return self.dyn1 ** self.dyn2


obj = Example(2, 8)

print("Метод 1:")
obj.method1()

print("Метод 2 (сумма глобальных a и b):", obj.method2())
print("Метод 3 (dyn1 ** dyn2):", obj.method3())
print("Переменная a:", a)
```

## Задание 2
Класс `Country`: столица, площадь, численность населения.  
Список объектов; вывод по площади и по численности.

```python
class Country:
    def __init__(self, name, capital, area, population):
        self.name = name
        self.capital = capital
        self.area = area          # км²
        self.population = population  # чел.

    def __str__(self):
        return (
            f"{self.name}: столица — {self.capital}, "
            f"площадь — {self.area} км², "
            f"население — {self.population} чел."
        )


countries = [
    Country("Россия", "Москва", 17_098_246, 146_000_000),
    Country("Франция", "Париж", 643_801, 67_000_000),
    Country("Япония", "Токио", 377_975, 125_000_000),
    Country("Беларусь", "Минск", 207_600, 9_200_000),
    Country("Казахстан", "Астана", 2_724_900, 19_000_000),
]


def by_area(country_list, min_area):
    """Страны с площадью >= заданной."""
    return [c for c in country_list if c.area >= min_area]


def by_population(country_list, min_population):
    """Страны с населением >= заданного."""
    return [c for c in country_list if c.population >= min_population]


area_limit = 500_000
pop_limit = 50_000_000

print(f"a) Страны с площадью >= {area_limit} км²:")
for c in by_area(countries, area_limit):
    print(" ", c)

print(f"\nb) Страны с населением >= {pop_limit} чел.:")
for c in by_population(countries, pop_limit):
    print(" ", c)
```

## Задание 3
Класс `Stationery` и дочерние: `Pen`, `Pencil`, `Handle` с переопределением `draw`.

```python
class Stationery:
    def __init__(self, title):
        self.title = title

    def draw(self):
        print("Запуск отрисовки")


class Pen(Stationery):
    def draw(self):
        print(f"Ручка «{self.title}»: рисуем чернилами")


class Pencil(Stationery):
    def draw(self):
        print(f"Карандаш «{self.title}»: рисуем графитом")


class Handle(Stationery):
    def draw(self):
        print(f"Маркер «{self.title}»: рисуем ярким цветом")


base = Stationery("Базовый инструмент")
pen = Pen("Parker")
pencil = Pencil("Koh-i-Noor")
handle = Handle("Edding")

print("Stationery:")
base.draw()

print("\nPen:")
pen.draw()

print("\nPencil:")
pencil.draw()

print("\nHandle:")
handle.draw()
```

## Задание 4
Собственный класс с методами экземпляра, статическими методами и методами класса.

Класс `Car` — учёт автомобилей и средний расход топлива.

```python
class Car:
    fleet_count = 0  # сколько машин создано

    def __init__(self, brand, model, fuel_consumption):
        self.brand = brand
        self.model = model
        self.fuel_consumption = fuel_consumption  # л / 100 км
        Car.fleet_count += 1

    # --- метод экземпляра ---
    def info(self):
        return f"{self.brand} {self.model}, расход {self.fuel_consumption} л/100 км"

    def cost_for_distance(self, km, price_per_liter):
        """Стоимость поездки на заданное расстояние."""
        liters = self.fuel_consumption * km / 100
        return liters * price_per_liter

    # --- статический метод ---
    @staticmethod
    def is_economical(consumption):
        """Экономичный ли расход (без доступа к self / cls)."""
        return consumption <= 7.0

    # --- метод класса ---
    @classmethod
    def from_string(cls, data):
        """Создаёт объект из строки 'Brand;Model;Consumption'."""
        brand, model, consumption = data.split(";")
        return cls(brand.strip(), model.strip(), float(consumption))

    @classmethod
    def get_fleet_count(cls):
        return cls.fleet_count


car1 = Car("Toyota", "Corolla", 6.5)
car2 = Car.from_string("BMW; X5; 11.2")

print("Авто 1:", car1.info())
print("Авто 2:", car2.info())
print(f"Поездка 200 км на {car1.model}: {car1.cost_for_distance(200, 55):.2f} руб.")
print("Corolla экономичная?", Car.is_economical(car1.fuel_consumption))
print("X5 экономичный?", Car.is_economical(car2.fuel_consumption))
print("Всего автомобилей в парке:", Car.get_fleet_count())
```
