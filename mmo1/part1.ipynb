## Задание 1
Вывести на экран все простые числа в заданном диапазоне (диапазон вводится с клавиатуры).

```python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True
    

a = int(input("Введите начало диапазона: "))
b = int(input("Введите конец диапазона: "))

if a > b:
    a, b = b, a

primes = [x for x in range(a, b + 1) if is_prime(x)]
print("Простые числа в диапазоне:", primes if primes else "нет")
```

## Задание 2
С клавиатуры вводится текст. Определить, сколько в нём гласных, а сколько согласных.  
В случае равенства вывести на экран все гласные буквы. Посчитать количество слов в тексте.

```python
vowels_set = set("аеёиоуыэюяaeiouy")
consonants_set = set("бвгджзйклмнпрстфхцчшщbcdfghjklmnpqrstvwxz")

text = input("Введите текст: ")

vowels_count = 0
consonants_count = 0
found_vowels = []

for ch in text.lower():
    if ch in vowels_set:
        vowels_count += 1
        found_vowels.append(ch)
    elif ch in consonants_set:
        consonants_count += 1

words = text.split()
words_count = len(words)

print(f"Гласных: {vowels_count}")
print(f"Согласных: {consonants_count}")
print(f"Количество слов: {words_count}")

if vowels_count == consonants_count:
    print("Гласные буквы:", "".join(found_vowels))
```

## Задание 3
Дан список чисел. Посчитайте, сколько в нем пар элементов, равных друг другу.  
Любые два элемента, равные друг другу, образуют одну пару.

```python
from collections import Counter

numbers = [1, 2, 3, 2, 1, 1, 4, 5, 4, 2]
print("Список:", numbers)

pairs = 0
for count in Counter(numbers).values():
    pairs += count * (count - 1) // 2

print("Количество пар равных элементов:", pairs)
```

## Задание 4
Дана строка — случайная последовательность чисел от 0 до 9.  
Создать словарь: ключи — числа (`int`), значения — количество вхождений.

```python
import random

sequence = "".join(str(random.randint(0, 9)) for _ in range(20))
print("Последовательность:", sequence)

digit_count = {}
for ch in sequence:
    key = int(ch)
    digit_count[key] = digit_count.get(key, 0) + 1

print("Словарь:", digit_count)
```

## Задание 5
Программа «Магазин автозапчастей» с меню:
1. Просмотр описания
2. Просмотр цены
3. Просмотр количества
4. Вся информация
5. Покупка
6. До свидания

```python
shop = {
    "Фильтр масляный": ["Очищает масло двигателя", 850, 15],
    "Свеча зажигания": ["Воспламенение топливной смеси", 320, 40],
    "Тормозная колодка": ["Комплект передних колодок", 2100, 8],
    "Аккумулятор": ["12В, 60 А·ч", 5500, 5],
    "Ремень ГРМ": ["Ремень газораспределительного механизма", 1800, 12],
}


def show_description():
    print("\n--- Описание ---")
    for name, info in shop.items():
        print(f"{name} — {info[0]}")


def show_price():
    print("\n--- Цены ---")
    for name, info in shop.items():
        print(f"{name} — {info[1]} руб.")


def show_quantity():
    print("\n--- Количество ---")
    for name, info in shop.items():
        print(f"{name} — {info[2]} шт.")


def show_all():
    print("\n--- Вся информация ---")
    for name, info in shop.items():
        print(f"{name}: описание — {info[0]}; цена — {info[1]} руб.; количество — {info[2]} шт.")


def buy():
    total = 0
    print("\n--- Покупка (n — выход) ---")
    while True:
        name = input("Название продукции: ").strip()
        if name.lower() == "n":
            break
        if name not in shop:
            print("Такого товара нет.")
            continue
        try:
            qty = int(input("Количество: "))
        except ValueError:
            print("Введите целое число.")
            continue
        if qty <= 0:
            print("Количество должно быть больше 0.")
            continue
        if qty > shop[name][2]:
            print(f"Недостаточно товара. В наличии: {shop[name][2]} шт.")
            continue
        cost = shop[name][1] * qty
        shop[name][2] -= qty
        total += cost
        print(f"Куплено: {name} × {qty} = {cost} руб. Осталось: {shop[name][2]} шт.")

    print(f"\nИтого к оплате: {total} руб.")
    print("Остатки на складе:")
    for name, info in shop.items():
        print(f"  {name}: {info[2]} шт.")


while True:
    print("\n===== Магазин автозапчастей =====")
    print("1. Просмотр описания")
    print("2. Просмотр цены")
    print("3. Просмотр количества")
    print("4. Вся информация")
    print("5. Покупка")
    print("6. До свидания")

    choice = input("Выберите пункт меню: ").strip()

    if choice == "1":
        show_description()
    elif choice == "2":
        show_price()
    elif choice == "3":
        show_quantity()
    elif choice == "4":
        show_all()
    elif choice == "5":
        buy()
    elif choice == "6":
        print("До свидания!")
        break
    else:
        print("Неверный пункт меню.")
```

## Задание 6
Дан кортеж целых чисел. Найти индекс первого отрицательного элемента.

```python
numbers = (5, 3, 0, 8, -2, 4, -7, 1)
print("Кортеж:", numbers)

index = None
for i, value in enumerate(numbers):
    if value < 0:
        index = i
        break

if index is not None:
    print(f"Индекс первого отрицательного элемента: {index} (значение {numbers[index]})")
else:
    print("Отрицательных элементов нет.")
```
